```python
import pandas as pd

```

```python
# sheet_name=None lee TODAS las hojas automáticamente
dfs = pd.read_excel("C:/Users/ingsoporte3/OneDrive - Syspotec/Documentos/empresas/Empresaas.xlsx"
                    ,sheet_name=None)     
```

```python
# Creamos tabla con DataFrame con el resumen de las hojas
resumen_hojas = pd.DataFrame({
    
    # Columna 1: Nombre de cada hoja
    "Nombre de la hoja": list(dfs.keys()),          # dfs.keys() = nombres de las hojas
    
    # Columna 2: Cantidad de filas de cada hoja
    "Cantidad de filas": [len(df) for df in dfs.values()]  
    # Explicación de la línea de arriba:
    # 1. dfs.values()     → todas las tablas
    # 2. for df in ...    → para cada tabla
    # 3. len(df)          → cuántas filas tiene
    # 4. [ ... ]          → guarda todo en una lista
})
resumen_hojas
```

```python
[len(df) for df in dfs.values()]  
```

**JUNTAR TODAS LAS HOJAS EN UN SOLO DATAFRAME (CONCATENACIÓN)**


```python
# pd.concat() es la función más sencilla para pegar DataFrames uno debajo de otro
df_total = pd.concat(
    dfs.values(),           # dfs.values() = todas las hojas (todas las tablas)
    ignore_index=True       # reinicia el índice para que empiece desde 0,1,2... sin duplicados
)

print(f" ¡Listo! Ahora tenemos UN SOLO DataFrame con {len(df_total):,} filas")
print(f"   Columnas originales: {len(df_total.columns)}")

# Mostramos las primeras 3 filas para verificar rápidamente
df_total.head(3)
```

```python
df_total
```

**Comenzamos limpieza, quitamos na, borramos espacios**

```python

#  Convertir la columna 'Date' a formato fecha (para poder filtrar por fechas)
# ---------------------------------------------------------------------
print("    Convirtiendo columna 'Date' a fecha...")

# pd.to_datetime() es la forma más sencilla de convertir texto a fecha
df_total['Date'] = pd.to_datetime(df_total['Date'], errors='coerce')

# ---------------------------------------------------------------------
# Convertir columnas numéricas a tipo número (Subtotal, Total, etc.)
# ---------------------------------------------------------------------
print("  Convirtiendo columnas numéricas...")

# Lista sencilla de columnas que deberían ser números
columnas_numericas = ['Subtotal', 'TaxSale', 'Total', 'Retention', 'Net', 'Brute']

for col in columnas_numericas:
    if col in df_total.columns:
        # pd.to_numeric() convierte texto a número y pone NaN donde no pueda
        df_total[col] = pd.to_numeric(df_total[col], errors='coerce')
        print(f"      ✓ Columna '{col}' convertida a número")

# ---------------------------------------------------------------------
# Información final después de la limpieza
# ---------------

print(f"   Total filas: {len(df_total):,}")
print(f"   Total columnas: {len(df_total.columns)}")
print("Tipos de datos actuales:")
print(df_total.dtypes)

# Mostrar las primeras 3 filas para ver el resultado
df_total.head(8)
```

```python
df_total.columns
```

```python
# =============================================================================
# PREPARACIÓN PARA VISUALIZACIÓN - Crear columnas de tiempo (muy sencillo)
# =============================================================================

print(" Creando columnas de Mes y Año para los KPIs...")

# Extraemos el año y el mes de la columna Date (que ya convertimos a fecha antes)
df_total['Año'] = df_total['Date'].dt.year
df_total['Mes'] = df_total['Date'].dt.month
df_total['Año_Mes'] = df_total['Date'].dt.strftime('%Y-%m')   # formato 2025-04

print("   Columnas creadas: 'Año', 'Mes' y 'Año_Mes'")
print(df_total[['Date', 'Año_Mes']].head(3))
```

```python
# =============================================================================
# KPI 1 - Cliente con mayor valor total
# =============================================================================

print(" Calculando cliente con mayor valor...")

# Creamos la columna de valor total exactamente como pediste
df_total['Valor_Total'] = df_total['Subtotal'] + df_total['TaxSale']

# Agrupamos por cliente y sumamos su valor total
valor_por_cliente = df_total.groupby('ClientName')['Valor_Total'].sum().sort_values(ascending=False)

# Obtenemos el cliente #1 y su valor
cliente_top = valor_por_cliente.index[0]
valor_top = valor_por_cliente.iloc[0]

print(f" CLIENTE CON MAYOR VALOR:")
print(f"   Cliente: {cliente_top}")
print(f"   Valor total: ${valor_top:,.2f}")
```

```python
# =============================================================================
# 9.  Mes con mayor valor del cliente top
# =============================================================================

print("Buscando el mes con mayor valor de ese cliente...")

# Filtramos solo las filas del cliente top
df_top_cliente = df_total[df_total['ClientName'] == cliente_top]

# Agrupamos por mes y sumamos
valor_por_mes = df_top_cliente.groupby('Año_Mes')['Valor_Total'].sum().sort_values(ascending=False)

# Mes con el pico más alto
mes_top = valor_por_mes.index[0]
valor_mes_top = valor_por_mes.iloc[0]

print(f"📅 MES CON MAYOR VALOR DEL CLIENTE {cliente_top}:")
print(f"   Mes: {mes_top}")
print(f"   Valor en ese mes: ${valor_mes_top:,.2f}")
```
